# 📊 Customer Churn Prediction: Exploratory Data Analysis (EDA)
**Goal:** Understand customer behavior, uncover patterns driving churn, and identify the most influential factors before modeling.

### Business Context:
Customer churn directly hurts Recurring Revenue (MRR/ARR) and Customer Lifetime Value (LTV). Acquiring a new customer costs 5x to 7x more than retaining an existing customer. By uncovering high-risk behavioral patterns, retention teams can target proactive campaigns.

```
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configure visual styling
sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 11
```

## 1. Load and Inspect Dataset

```
data_path = '../data/raw/telco_customer_churn.csv'
df = pd.read_csv(data_path)
print(f'Shape of dataset: {df.shape}')
df.head()
```

## 2. Data Cleaning & Type Conversion
`TotalCharges` contains blank spaces when `tenure == 0`. We convert it to numeric and impute 0.0 for those records.

```
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'].replace(' ', np.nan), errors='coerce')
df['TotalCharges'] = df['TotalCharges'].fillna(0.0)
print(df.info())
```

## 3. Churn Class Distribution Analysis

```
churn_counts = df['Churn'].value_counts()
print(churn_counts)
print(f'Churn Rate: {(churn_counts["Yes"] / len(df)) * 100:.2f}%')

fig, ax = plt.subplots(figsize=(6, 4))
sns.countplot(data=df, x='Churn', hue='Churn', palette=['#2b5c8f', '#d9534f'], legend=False, ax=ax)
ax.set_title('Churn Class Distribution', fontsize=14, fontweight='bold')
plt.show()
```

## 4. Bivariate Analysis: Churn by Contract Type & Billing

```
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
contract_churn = df.groupby('Contract')['Churn'].apply(lambda x: (x == 'Yes').mean() * 100).reset_index()
sns.barplot(data=contract_churn, x='Contract', y='Churn', hue='Contract', palette='Reds_r', legend=False, ax=ax1)
ax1.set_title('Churn Rate by Contract Type (%)', fontweight='bold')
ax1.set_ylabel('Churn Rate (%)')

pay_churn = df.groupby('PaymentMethod')['Churn'].apply(lambda x: (x == 'Yes').mean() * 100).reset_index()
sns.barplot(data=pay_churn, x='PaymentMethod', y='Churn', hue='PaymentMethod', palette='Blues_r', legend=False, ax=ax2)
ax2.set_title('Churn Rate by Payment Method (%)', fontweight='bold')
ax2.set_ylabel('Churn Rate (%)')
ax2.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()
```

## 5. Tenure and Charges Distribution

```
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
sns.kdeplot(data=df, x='tenure', hue='Churn', common_norm=False, fill=True, palette={'No': '#2b5c8f', 'Yes': '#d9534f'}, ax=ax1)
ax1.set_title('Tenure Distribution by Churn', fontweight='bold')
ax1.set_xlabel('Tenure (Months)')

sns.kdeplot(data=df, x='MonthlyCharges', hue='Churn', common_norm=False, fill=True, palette={'No': '#2b5c8f', 'Yes': '#d9534f'}, ax=ax2)
ax2.set_title('Monthly Charges Distribution by Churn', fontweight='bold')
ax2.set_xlabel('Monthly Charges ($)')
plt.tight_layout()
plt.show()
```

## 6. Services & Tech Support Impact

```
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
ts_churn = df.groupby('TechSupport')['Churn'].apply(lambda x: (x == 'Yes').mean() * 100).reset_index()
sns.barplot(data=ts_churn, x='TechSupport', y='Churn', hue='TechSupport', palette='Purples_r', legend=False, ax=ax1)
ax1.set_title('Churn Rate by Tech Support Status (%)', fontweight='bold')

sec_churn = df.groupby('OnlineSecurity')['Churn'].apply(lambda x: (x == 'Yes').mean() * 100).reset_index()
sns.barplot(data=sec_churn, x='OnlineSecurity', y='Churn', hue='OnlineSecurity', palette='Greens_r', legend=False, ax=ax2)
ax2.set_title('Churn Rate by Online Security Status (%)', fontweight='bold')
plt.tight_layout()
plt.show()
```

## 7. Key Findings & Business Takeaways
1. **Contract Friction:** Month-to-month customers churn at 42.7% versus 2.8% for 2-year customers.
2. **Critical Retention Window:** The first 1-12 months are high-risk onboarding phases.
3. **Protective Add-ons:** Tech Support and Online Security cut customer churn by more than half (41.6% down to 15.2%).
